# KNN imputation

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

# 1. Criar um DataFrame de exemplo (substitua pelo seu)
data = {
    'Timestamp': pd.to_datetime(pd.date_range(start='2023-01-01', periods=100, freq='T')),
    'Var_A': np.linspace(10, 20, 100) + np.random.randn(100) * 0.5,
    'Var_B': np.linspace(50, 40, 100) + np.random.randn(100) * 0.5,
    'Var_C': np.linspace(1, 2, 100) + np.random.randn(100) * 0.1
}
df = pd.DataFrame(data)

# Introduzir valores faltantes
for col in ['Var_A', 'Var_B', 'Var_C']:
    df.loc[df.sample(frac=0.1).index, col] = np.nan

df.set_index('Timestamp', inplace=True)
print("DataFrame Original com Valores Faltantes:")
print(df.head())
print("\nContagem de NaN antes da imputação:")
print(df.isnull().sum())

# 2. Aplicar o KNNImputer
# O imputer funciona melhor com dados numéricos, por isso separamos o timestamp
imputer_knn = KNNImputer(n_neighbors=5, weights='uniform')

# fit_transform retorna um array NumPy, então precisamos reconstruir o DataFrame
df_imputado_knn_array = imputer_knn.fit_transform(df)
df_imputado_knn = pd.DataFrame(df_imputado_knn_array, columns=df.columns, index=df.index)

print("\nDataFrame Imputado com KNN:")
print(df_imputado_knn.head())
print("\nContagem de NaN após a imputação com KNN:")
print(df_imputado_knn.isnull().sum())
```

# Iteractive imputation

```python
# É necessário importar enable_iterative_imputer para usar a funcionalidade experimental
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.linear_model import BayesianRidge # Um regressor comum para isso

# Usando o mesmo df criado no exemplo anterior

# 3. Aplicar o IterativeImputer
# O estimador padrão é BayesianRidge, que é uma boa escolha geral.
# max_iter define o número de ciclos de imputação.
imputer_iterative = IterativeImputer(max_iter=10, random_state=42)

df_imputado_iter_array = imputer_iterative.fit_transform(df)
df_imputado_iter = pd.DataFrame(df_imputado_iter_array, columns=df.columns, index=df.index)

print("\nDataFrame Imputado com IterativeImputer (MICE):")
print(df_imputado_iter.head())
print("\nContagem de NaN após a imputação com IterativeImputer:")
print(df_imputado_iter.isnull().sum())
```
